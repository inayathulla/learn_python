# Exception Handling

### Compile time errors
Mistakes in syntax are basically **_compile_** time errors.
You will come to know about these errors before you run the program, as program will not run otherwise.

Ex:- While printing giving **_P_** as capital as in **_Print_** instead **_print_**

```python
print('Hello') # Compile time error when p is capital
```

### Runtime errors
Here syntax is 100% correct. And these errors are caused at the runtime. And if we do not handle it then program will crash abruptly.

Ex:- IndexError: list index out of range

```python
l = [1, 2, 3]
index = int(input('Enter index number of element: '))
print(l[index])
```

And the errors which are handled are called **_exceptions_**.

Below is one more classical example, which throws **_ZeroDivisionError: division by zero_**

```python
x = 10
y = 0
print(x / y)
```

The idea is to identify the potential code which can throw errors and handle it with **_try - except - finally_**

#### Simple **_try except_**

```python
x = 20
y = 0
try:
    print(x / y)
except Exception as e:
    print('Exception caught: ',e)
print('Program continued after catching exception')
```

#### One **_try_** multiple **_except_**

```python
index = int(input('Enter list element index to print: '))
x = 30
try:
    print(f'Element at index {index} is: ',l[index])
    print('Printing awesome division..')
    print(x / y)

except IndexError as e:
    print('Exception caught: ', e)
except ZeroDivisionError as e:
    print('Exception caught: ', e)
print('Program handled exceptions')
```

#### **_try except finally_**
If you want to close resources properly irrespective of normal or exception flow then we will keep it inside **_finally_**

```python
try:
    num = int(input("Enter a number: "))
    print(10 / num)

except ValueError:
    print("Please enter a valid number.")

except ZeroDivisionError:
    print("Cannot divide by zero.")

finally:
    print("Program finished.")
```

#### Nested Exception

```python
try:
    num = int(input("Enter a number: "))
    try:
        result = 10 / num
        print("Result:", result)
    except ZeroDivisionError:
        print("Inner Exception: Cannot divide by zero.")
except ValueError:
    print("Outer Exception: Please enter a valid number.")
```

#### Custom Exception with **_raise_** keyword

```python
class AgeNotEligibleException(Exception):
    pass

age = int(input("Enter age: "))
try:
    if age < 18:
        raise AgeNotEligibleException("Age must be 18 or above.")

    print("Eligible")

except AgeNotEligibleException as e:
    print(e)
```
