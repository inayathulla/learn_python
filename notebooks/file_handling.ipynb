# File Handling

In order to perform any operations on file, we should open it first and then should close in the end.

You will **_Open_** file by passing it 2 parameters.
1. File Name
2. Mode (r-read, a-append, w-write, x-create)

* Read will be used to read contents and if the file doesn't exists then it will raise **_FileNotFoundError_**.
* Append will add the contents at the end of file.
* Write will override the contents if file is present else create new file and write the contents.
* Create is used when you want to create the file and write content and if the file already exists then it will raise **_FileExistsError_**.

* **Note:** **_PermissionError_** is raised when access-denied to read or write at location with no access.

#### Reading

```python
file = open('../data/sample.txt', 'r')

print(file.read(4))
print(file.readline()) # Read one line at a time
print(file.read()) # Read all lines at once
```

#### Writing

```python
file = open('../data/sample_w.txt', 'w')

file.write("This is first line\n")
file.write("This is second line")
file.write("This is third line\n")

file = open('../data/sample_w.txt', 'r')
print(file.read())
```

```python
file = open('../data/sample_w.txt', 'a')
file.write("This is fourth line\n")

file = open('../data/sample_w.txt', 'r')
print(file.read())
```

We close the file finally as **_file.close()_**
