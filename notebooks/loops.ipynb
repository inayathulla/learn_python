# Loops

# While

If you want to execute code block as long as condition is true.

```python
a = 5
while a < 10:
    print(a)
    a += 1
```

# for

If you want to execute code block N number of times

```python
for i in range(5):
    print(i, end=" ")
```

```python
for i in range(3, 10):
    print(i, end="-")
```

```python
for i in range(1, 10, 2):
    print(i, end=",")
```

```python
for i in "Star":
    print(i, end="=")
```

# break and continue keyword

```python
j = 5
while j > 0:
    if j == 2:
        break
    print(j)
    j -= 1
```

```python
k = 0
while k <= 5:
    if k == 2:
        k += 1
        continue
    print(k)
    k += 1
```
