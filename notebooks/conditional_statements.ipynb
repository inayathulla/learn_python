# Conditional Statements

### if condition
_Indentation_ is strictly followed. Statements only following corresponding indentation will be executed.

```python
num = int(input('Enter a number: '))

if num == 5:
    print('Correct!')
print('This is outside of if condition')
```

### else condition

```python
num = int(input('Enter a number: '))

if num == 5:
    print('Correct!')
else:
    print('Incorrect!')
```

### elif condition
If multiple conditions needs to be checked then elif is handy


```python
marks = int(input('Enter a marks: '))

if marks > 70:
    print('Passed with Distinction!')
elif marks > 60:
    print('Passed with First Class!')
elif 40 < marks < 60:
    print('Passed!')
else:
    print('Failed!')
```

### Nesting Conditions

```python
a = 10
if a > 4:
    if a < 10:
        print('a is between 5 and 10')
    else:
        print('a greater than 10')
```
